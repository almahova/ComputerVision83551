
## Inline Question 2:

Now that you have trained a Neural Network classifier, you may find that your testing accuracy is much lower than the training accuracy. In what ways can we decrease this gap? Select all that apply.

1. Train on a larger dataset.
2. Add more hidden units.
3. Increase the regularization strength.
4. None of the above.

$\color{blue}{\textit Your Answer:}$ 1 and 3

$\color{blue}{\textit Your Explanation:}$<br/>
Train on a larger dataset: yes. More varied examples make it harder for the network to memorize individual images. It has to learn patterns that hold across the data, so test accuracy moves closer to training accuracy.<br/>
Add more hidden units: no. More hidden units give the model more capacity, which makes it easier to fit noise in the training set. That usually widens the gap. It only helps when the model is underfitting, which is a different problem.<br/>
Increase the regularization strength: yes. A stronger L2 penalty keeps the weights small and the learned function smoother. This limits how closely the model can fit the training data, which narrows the gap. Too much regularization will cause underfitting, though.
